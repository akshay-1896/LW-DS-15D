```python
def greet(name1,name2):
    print(f"Hello {name1} & {name2}, welcome to python!")
```

```python
greet("Kunal","Akshay")
```

```python
greet(name2="Kunal",name1="Akshay")
```

```python
def introduction(name, age):
    print(f'Hello, I am {name}, and I am {age} years old')
    # print('Hello, I am',name,'and I am ,' age,' years old')
```

```python
introduction(age=21,name="Kunal") # Keyword arguments
```
