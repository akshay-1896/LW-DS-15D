```python
import numpy as np
import pandas as pd
```

## **Merging**

```python
# DF1: Employee Information
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Clever', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DF2: Salary Information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees,salaries)
```

```python
pd.merge(employees,salaries,how='outer',on='employee_id') # considers all elements
```

```python
pd.merge(employees,salaries,how='left')# w.r.t left dataframe
```

```python
pd.merge(employees,salaries,how='right') # w.r.t right dataframe
```

## **Concatenation**

```python
df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})

df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1,df2]) # on the basis columns
```

```python
pd.concat([df1,df2],axis=1) # on the basis of rows
```

## **Joining**

```python
df1 = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie']
}, index=[1, 2, 3])

# Second DataFrame
df2 = pd.DataFrame({
    'score': [85, 90, 75]
}, index=[2, 3, 4])
```

```python
df = pd.DataFrame()
```

```python
df1
```

```python
df2
```

```python
df1.join(df2,how='outer')
```

```python
df2.join(df1,how='outer')
```

```python
df1.join(df2,how='inner')
```

```python
df1.join(df2,how='left')
```

```python
df1.join(df2,how='right')
```
