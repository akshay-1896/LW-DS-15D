### Creating Dataframe
- Multiple series combined are termed as Dataframe

```python
import numpy as np
import pandas as pd
```

#### From Dictionary

```python
data = {
    'Name': ['John', 'Anna', 'Peter', 'Clever'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}
df = pd.DataFrame(data)
```

```python
df
```

#### From List

```python
# When creating df from list, in list headers or column names are not mentioned, therefore by default every column is mentioned with 0,1,2,3,etc.

data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Clever', 42, 'London', 85000]
]
df2 = pd.DataFrame(data_list)
cols = ['Name','Age','City','Salary']
df2 = pd.DataFrame(data_list,columns=cols) # This mentions columns
```

```python
df2
```

### Selection and Indexing of Columns

```python
df2
```

```python
# df2[0] --> Throws keyerror
df2['Name']
df2['Age']
```

```python
df2[['Name','City']] # To get multiple columns, we provide list of columns
```

### Creating a new column

```python
# Since there doesn't exist any Designation column, so it will create one and assign respective values

# Whenever we create column in this format, we have to pass exact no. of values in the list as there are no. of elements listed i.e. 4, otherwise it will throw error 

df2['Designation'] = ['Doctor', 'ML Engineer', 'Teacher', 'Doctor']
```

```python
df2
```

### Removing any Column

```python
# df2.drop('Designation') # Throws error --> "['Designation'] not found in axis"
df2.drop('Designation', axis=1) 
# By default Axis=0 searches out in row indices vertically --> horizontal drops 
# Axis=1 searches out in Column indices horizontally --> vertical drops
```

```python
df2 # inplace=False --> Designation does not gets removed
```

```python
df2.drop('Designation', axis=1, inplace=True) #--> This removes column permanently
df2
```

```python
# Removing multiple columns (vertical drops)
df2.drop(['City','Salary'], axis=1)
```

```python
# Removing row
# df2.drop('0') # Throws error -->  "['0'] not found in axis"
df2.drop(0)
```

### Conditional Selection

```python
df2
```

```python
# People with age > 30

df2[df2['Age'] > 30]
```

```python
# People with age > 30 & City is Paris

# df2[df2['Age'] > 30 & df2['City'] == 'Paris'] # Throws error because these conditions are compared with Parenthesis
df2[(df2['Age'] > 30) & (df2['City'] == 'Paris')]
```
