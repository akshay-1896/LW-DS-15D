## **Group By**
- Performed over categorical columns or qualitative columns

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
df = pd.DataFrame(data)
```

```python
df
```

```python
# Group By Category and calculate the sum of sales

cat = df.groupby(by='Category') # creates object which cannot be seen directly
cat # cannot be displayed directly
```

### **_Why GroupBy gives object but not dataframe representation with full data?_**

groupby() does not perform the grouping immediately. Instead, it creates a GroupBy object that stores the grouping instructions. This object waits until you specify what operation you want (such as sum(), mean(), count(), agg(), transform(), or filter()).

This approach is known as **lazy evaluation**.

**Why Doesn't Pandas Show the Data Immediately?**

- Because Pandas doesn't know what you want to do with the groups.

**What Does the GroupBy Object Actually Store?**

Internally, it stores information like:

- Original DataFrame
- Column(s) used for grouping
- Which rows belong to each group
- Metadata about the grouping

**Benefits of Creating a GroupBy Object**

- Grouping can be expensive for large datasets. 
- Instead of immediately calculating everything,
- Pandas waits until you ask.
- **Memory Efficiency**
    - The object stores references (like row indices) rather than duplicating the grouped data.

**Enables Split–Apply–Combine**

- groupby() is built around the classic Split–Apply–Combine strategy.
- The GroupBy object represents the split stage. Later methods perform the apply and combine stages.

```python
for i, v in cat:
    print(i)
    print(v)
```

```python
# Group by Category and calculate the sum of Sales
cat = df.groupby(by='Category')['Sales'].sum()
cat
```

```python
# Group by Store and calculate the sum of Sales
cat = df.groupby(by='Store')['Sales'].sum()
cat
```

```python
# Group by multiple columns
# Group by Category and Store
cat = df.groupby(['Category','Store'])['Sales'].sum()
cat
```
