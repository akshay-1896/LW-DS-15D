```python
import numpy as np
import pandas as pd
```

## **Pandas Series**

- A Series is a one-dimensional labeled array capable of holding any data type. The axis labels are collectively called the index. Array are horizontally formatted and Series are vertically formatted

```python
labels = ['a','b','c']
my_list = [10,20,30]
arr = np.array([10,20,30])
d = {1:10,2:20,3:30}
```

```python
s = pd.Series(my_list)
s
# type(s)
```

```python
pd.Series(my_list, index=labels)
```

```python
pd.Series(arr)
```

**Note:** Never pass 2-D array because Series itself are 1-D

```python
pd.Series(arr, index=labels)
```

```python
pd.Series(d)
```
