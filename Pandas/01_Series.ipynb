## *Pandas Series*
- A Series is a one-dimensional labeled array capable of holding any data type. The axis labels are collectively called the index. Array are horizontally formatted and Series are vertically formatted

### Creating Series

```python
import pandas as pd
import numpy as np
```

```python
labels = ['a','b','c']
my_list = [10,20,30]
arr = np.array([10,20,30])
d = {1:10,2:30,3:30}
```

```python
s = pd.Series(my_list)
s
type(s)
```

```python
pd.Series(my_list, index=labels)
```

```python
pd.Series(arr)
```

```python
pd.Series(arr, index=labels) # Never pass 2-D array because Series itself are 1-D
```

```python
pd.Series(d) # In the case of dictionary, there is no need for typecasting indices because by default all values are indexed with respect to their keys
```
