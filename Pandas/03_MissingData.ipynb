```python
import numpy as np
import pandas as pd
```

### Finding Missing Data

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any() # Which colums have Null values
```

### Removing Missing Data

```python
df
```

```python
# Remember: Removing missing data works on the basis of rows

df.dropna()
```

```python
df
```

```python
df.dropna(thresh=3) # It provides those rows which have minimum non-null values w.r.t defined threshold i.e. 3
```

```python
df.dropna(thresh=4)
```

### Filling the Missing Data

```python
df
```

```python
df.fillna(0)
```

```python
# Filling different null values for different columns 

values = {'A':0,'B':100,'C':'300','D':400}
df.fillna(value=values)
```

```python
df
```

```python
# Fills mean values in place of Null values

df.fillna(df.mean())
```
