## **Group By**

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
df = pd.DataFrame(data)
```

```python
df
```

```python
cat = df.groupby(by='Category')
cat
# sum(),mean(),count()
```

```python
for i, v in cat:
    print(i)
    print(v)
```

```python
cat_sum = df.groupby(by='Category')['Sales'].sum()
cat_sum
```

```python
cat_sum = df.groupby(by='Store')
cat_sum
```

```python
for i, v in cat_sum:
    print(i)
    print(v)
```

```python
cat = df.groupby(by = ['Category', 'Store'])['Sales'].sum()
cat
```
