## **Pivot Tables**

A pivot table summarizes data by organizing one or more categorical variables into rows and columns and applying an aggregation function to numerical data.
- **"Group my data by these row and column categories, then calculate something."**
- Used in creation of heatmaps
- Aggregation Functions are used in operation or calculation
- Aggregation-oriented Summary

```python
data = {
    'Date': pd.date_range('2023-01-01', periods=20),
    'Product': ['A', 'B', 'C', 'D'] * 5,
    'Region': ['East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West',
               'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    'Sales': np.random.randint(100, 1000, 20),
    'Units': np.random.randint(10, 100, 20),
    'Rep': ['John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary',
            'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice']
}

df = pd.DataFrame(data)
```

```python
df['Month'] = df['Date'].dt.month_name()
df.head()
```

```python
df['Quarter'] = 'Q' + df['Date'].dt.quarter.astype(str)
df.head()
```

**Note:** In a function definition, wherever there are parameters marked as None are required to be filled

```python
pd.pivot_table(df,values='Sales',index='Region',columns='Product')
```

```python
pivot1 = pd.pivot_table(df,values='Sales',index='Region',columns='Product',aggfunc='median')
pivot1
```

```python
pivot2 = pd.pivot_table(df,values=['Sales','Units'],index='Region',columns='Product')
pivot2
```

```python
# This gives error because Pivot Tables uses aggfunc() in operation where only numerical data is used
pivot2 = pd.pivot_table(df,values=['Sales','Quarter'],index='Region',columns='Product')
pivot2
```

## **Cross Tabs**

A crosstab, short for cross-tabulation, is primarily used to calculate the frequency or count of combinations of categorical variables.
- **"Tell me how often these two categories occur together."**
- Counting is used in calculation
- Frequency-oriented Summary

```python
df.head()
```

```python
pd.crosstab(df['Region'],df['Product'])

# The key distinction is: df[['Region', 'Product']] means "select these two columns," whereas pd.crosstab(df['Region'], df['Product']) means "use Region as the row variable and Product as the column variable."
# df['Region']   # First Series → used as rows
# df['Product']  # Second Series → used as columns
```
