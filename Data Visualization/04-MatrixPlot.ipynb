#### **Correlation**
- Correlation is a statistical measure that tells us how strongly and in what direction two variables are related.
- Types:
    - Positive Correlation: As the hours studied increase, the exam score also increases.
    - Negative Correlation: As speed increases, the time taken decreases.

**Correlation Coefficient**

The strength of correlation is measured by the correlation coefficient, usually denoted by r.

The value of r always lies between: `−1 ≤ r ≤ 1`

```python
tipscorr = tips[['total_bill','tip','size']]
tipscorr
```

```python
tipscorr.corr()
```

```python
# sns.heatmap(tipscorr.corr(),annot=True)
sns.heatmap(tipscorr.corr(),annot=True,cmap='crest')
```

### **Cluster Map**

```python
sns.clustermap(tipscorr.corr()) # Uses Heirarchical Algorithm
```
