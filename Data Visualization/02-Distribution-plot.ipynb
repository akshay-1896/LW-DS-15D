## **Distribution Plots**

- A distribution describes how the values of a variable are spread across a range.
- We use distribution plots to visualize the distribution of quantitative data.

- Types:
    - Displot/Histplot
    - Joint plot
    - Pair plot
    - Rug plot

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = sns.load_dataset('tips')
# type(sns.load_dataset('tips'))
df
```

```python
df['size'].unique()
```

### **Histogram Plot**

- It divides the data into intervals called bins and counts how many observations fall into each bin.
- Each bar shows how many observations fall within that interval.
- More bins give greater detail, while fewer bins provide a broader overview.
- In Normal Distribution, histogram forms Kernel Density Estimate which is called as Bell Curve.

**When to Use Distribution Plots**

- Distribution plots are commonly used during Exploratory Data Analysis (EDA) to:

    - Understand the shape of numerical variables.
    - Detect skewness and outliers.
    - Assess normality before applying statistical tests.
    - Compare distributions across different groups.
    - Guide feature engineering and preprocessing for machine learning.

```python
sns.displot(df['total_bill'],bins=6)
# sns.displot(df['total_bill'], bins= 100)
```

```python
sns.histplot(df['total_bill'],bins=100)
```

### **KDE (Kernel Density Estimation)**

Kernel Density Estimation (KDE) is a statistical technique used to estimate the probability density function (PDF) of a continuous variable. In simpler terms, it creates a smooth curve that represents how the data is distributed.

- **Kernel**: The small smooth function (usually Gaussian) placed at each data point.
- **Density**: It estimates the probability density of the data.
- **Estimation**: It approximates the true underlying distribution from the sample.

```python
sns.histplot(df['total_bill'],bins=100,kde=True)
```

```python
plt.subplot(1,2,1)
sns.histplot(df['total_bill'], bins = 20, kde=True)
plt.subplot(1,2,2)
sns.histplot(df['tip'], bins = 20, kde=True)
```
