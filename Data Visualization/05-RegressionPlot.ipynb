```python
import seaborn as sns
import matplotlib.pyplot as plt
```

## **Regression Plot**

- A regression plot (or regplot) is a visualization that combines a scatter plot with a regression line (line of best fit). It helps us understand whether there is a relationship between two numerical variables and how strong that relationship is.

```python
df = sns.load_dataset('tips')
df
```

```python
sns.lmplot(x = 'total_bill', y = 'tip', data=df)
```

```python
sns.lmplot(x = 'total_bill', y = 'tip', data=df, hue='smoker', palette='bright')
```

```python
sns.lmplot(x = 'total_bill', y = 'tip', data=df, hue='smoker',palette='YlGnBu', markers=['o','*'])
```
