# **Bag of Words**

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
documents = [
    "I love vadapav",
    "I love rasagulla",
    "I love maalpua",
    "Kaju katli is great",
    "Cham cham is the best"
]
```

```python
# vectorizer_BoW = CountVectorizer()

# vectorizer_BoW = CountVectorizer(ngram_range=(1,2))
# vectorizer_BoW = CountVectorizer(ngram_range=(2,3))
vectorizer_BoW = CountVectorizer(ngram_range=(3,3))
```

```python
X_bow = vectorizer_BoW.fit_transform(documents)
```

```python
print(f"Vocabulary: {vectorizer_BoW.get_feature_names_out()}")
print(f"\nBoW Matrix: \n{X_bow.toarray()}\n")
```
