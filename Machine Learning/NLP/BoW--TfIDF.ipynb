# **Bag of Words**

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
documents = [
    "I love pizza",
    "Pizza is the best",
    "I love pasta",
    "Pasta is great"
]
```

```python
vectorizer_BoW = CountVectorizer()

# N-grams
# vectorizer_BoW = CountVectorizer(ngram_range=(1,2))
# vectorizer_BoW = CountVectorizer(ngram_range=(2,2))
# vectorizer_BoW = CountVectorizer(ngram_range=(3,3))
# vectorizer_BoW = CountVectorizer(ngram_range=(1,3))
```

```python
X_bow = vectorizer_BoW.fit_transform(documents)
```

```python
print(f"Vocabulary: {vectorizer_BoW.get_feature_names_out()}")  #It is unordered
print(f"\nBoW Matrix: \n{X_bow.toarray()}\n")
```

# **TF - IDF**

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
vectorizer_tfidf = TfidfVectorizer()
```

```python
X_tfidf = vectorizer_tfidf.fit_transform(documents)
```

```python
print(f"Vocabulary: {vectorizer_tfidf.get_feature_names_out()}")  #It is unordered
print(f"\nTfidf Matrix: \n{X_tfidf.toarray()}\n")
```
