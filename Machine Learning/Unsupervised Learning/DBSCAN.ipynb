```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_moons
from sklearn.cluster import KMeans, DBSCAN
from sklearn.preprocessing import StandardScaler
```

```python
X, y_true = make_moons(n_samples=500, noise=0.05, random_state=42)
```

```python
df = pd.DataFrame(X, columns=['Feature_1', 'Feature_2'])
df.head()
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)
```

```python
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans_labels = kmeans.fit_predict(X_scaled)
```

```python
df['kmeans_cluster'] = kmeans_labels
df.head()
```

```python
sns.scatterplot(x=df['Feature_1'],y=df['Feature_2'],hue=df['kmeans_cluster'],palette='viridis')
```

```python
dbscan = DBSCAN(eps=0.3, min_samples=5)     # min_samples ----> for eps=0.3 take min-5 datapoints in your circle
dbscan_labels = dbscan.fit_predict(X_scaled)
```

```python
df['dbscan_cluster'] = dbscan_labels
df.head()
```

```python
sns.scatterplot(x=df['Feature_1'],y=df['Feature_2'],hue=df['dbscan_cluster'],palette='viridis')
```
