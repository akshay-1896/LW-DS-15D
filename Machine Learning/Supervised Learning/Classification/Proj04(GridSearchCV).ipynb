```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
df['species'].unique()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = df.drop('species',axis=1)
y = df['species']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

### **KNN**

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
# model_knn = KNeighborsClassifier(n_neighbors=5)
model_knn = KNeighborsClassifier(n_neighbors=13)
```

```python
model_knn.fit(X_train,y_train)
```

```python
model_knn.score(X_test,y_test)
# n = 5 >>>> A = 98%
# n = 13 >>>> A = 100% ----> Overfitted ----> This is prevented through Cross Validation
```

### **SVM**

```python
from sklearn.svm import SVC
```

```python
# model_svm = SVC(C=30,kernel='rbf',gamma='auto') 
# model_svm = SVC(C=10,kernel='linear',gamma='auto') 
model_svm = SVC(gamma='auto') # ----> For GridSearch CV
```

```python
model_svm.fit(X_train,y_train)
```

```python
model_svm.score(X_test,y_test)
```

# **Grid Search CV**

### **SVM**

```python
from sklearn.model_selection import GridSearchCV
```

```python
classifier = GridSearchCV(model_svm,{
    'C' : [1,10,20,30],
    'kernel' : ['rbf','linear'] 
},cv=5,return_train_score=False)
```

```python
classifier.fit(X,y) # We have not used X_train & y_train because CV will perform train-test-split by itself
```

```python
classifier.cv_results_
```

```python
results = pd.DataFrame(data=classifier.cv_results_)
```

```python
results
```

```python
results[['param_C','param_kernel','mean_test_score']]
```

#### **Perform GridSearchCV for KNN**

# **Random Search CV**

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
classifier_r = RandomizedSearchCV(model_svm,{
    'C' : [1,10,20,30],
    'kernel' : ['rbf','linear'] 
},n_iter = 4,cv=5,return_train_score=False) # ----> 4 iterations
```

```python
classifier_r.fit(X,y)
```

```python
results = pd.DataFrame(classifier_r.cv_results_)
```

```python
results[['param_C','param_kernel','mean_test_score']]
```
