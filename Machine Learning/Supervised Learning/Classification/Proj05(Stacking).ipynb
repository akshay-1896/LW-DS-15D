```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import StackingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report
from sklearn.preprocessing import LabelEncoder
import seaborn as sns
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
X = df.drop('species', axis=1)
y = df['species']
```

```python
le = LabelEncoder()
y_encoded = le.fit_transform(y)
```

# **Stacking Implementation**

```python
# In cross validation we do not perform train-test-split, but in ensemble techniques we have perform train-test-split

X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded)
```

```python
base_learners = [
    ('dt', DecisionTreeClassifier(random_state=42)),
    ('svc', SVC(probability=True, kernel='rbf', random_state=42)),
    ('logr', LogisticRegression(max_iter=1000))
]
```

```python
meta_learner = LogisticRegression(max_iter=1000)
```

```python
stacking_clf = StackingClassifier(
    estimators=base_learners,
    final_estimator=meta_learner,
    cv=5
)
```

```python
stacking_clf.fit(X_train,y_train)
```

```python
y_pred = stacking_clf.predict(X_test)
y_pred
```

```python
y_test
```

```python
accuracy = accuracy_score(y_test,y_pred)
accuracy
```

# **Bagging Implementation - Random Forest**

```python
from sklearn.ensemble import RandomForestClassifier 
```

```python
rf_model = RandomForestClassifier(
    n_estimators=100,   # number of trees
    max_depth=None,     # let trees grow fully
    random_state=42
)

rf_model.fit(X_train, y_train)
```

```python
y_pred_rf = rf_model.predict(X_test)
```

```python
accuracy = accuracy_score(y_test,y_pred_rf)
accuracy
```

# **Boosting Implementation**

## **Ada Boost**

```python
from sklearn.ensemble import AdaBoostClassifier
```

```python
ada_model = AdaBoostClassifier(n_estimators=100, random_state=42)
```

```python
ada_model.fit(X_train,y_train)
```

## **Gradient Boost**

```python
from sklearn.ensemble import GradientBoostingClassifier
```

```python
gb_model = GradientBoostingClassifier(
    n_estimators=100,   # Value of estimators(Decision Trees) depends upon capacity of our system
    learning_rate=0.1,
    random_state=42
)

gb_model.fit(X_train, y_train)
```

```python
y_pred_gb = gb_model.predict(X_test)
```

```python
accuracy = accuracy_score(y_test,y_pred_gb)
accuracy
```

## **XG Boost**

```python
from xgboost import XGBClassifier
```

```python
xgb_model = XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    use_label_encoder=False,
    eval_metric='mlogloss',
    random_state=42
)

xgb_model.fit(X_train,y_train)
```

```python
y_pred_xgb = xgb_model.predict(X_test)
```

```python
accuracy = accuracy_score(y_test,y_pred_xgb)
accuracy
```
