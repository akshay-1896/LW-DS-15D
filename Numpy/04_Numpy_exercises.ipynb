## *Valid Sudoku*
- A sudoku is valid iff all individual rows and columns have there individual sum as 45

```python
s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2], # For every row, the sum of the elements should be 45
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
```

```python
# s.sum()
r = np.sum(s,axis=1) # Each and every row has a sum of 45
r == 45
```

```python
for i in r:
    if i != 45:
        print("Sudoku is invalid")
        break

else:
    print("for rows Sudoku is valid")
```

```python
c = np.sum(s,axis=0)
c == 45
```

```python
for i in c:
    if i != 45:
        print("Sudoku is invalid")
        break

else:
    print("For columns Sudoku is valid")
```

```python
d1 = np.sum(s[0:3,0:3])
d1
```

```python
d2 = np.sum(s[0:3,3:6])
d2
```

```python
for i in range(0,9,3):
    for j in range(0,9,3):
        n = s[i:i+3,j:j+3]
        print(n.sum())
```

```python
l = list()
for i in range(0,9,3):
    for j in range(0,9,3):
        n = s[i:i+3,j:j+3]
        l.append(n.sum().item()) # n.sum() this gives Numpy scalar np.int64(45)
                                 # n.sum().item() gives Python scalar 45

l
```

## *General Questions*

```python
import numpy as np

# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

Get the shape of the matrix

```python
# Get the shape of the matrix
data.shape
```

Find the average age of students

```python
# Find the average age of students
np.mean(data[:,0])
```

Extract Math marks of all students

```python
# Extract Math marks of all students
data[:,1]
```

Find the highest Science marks

```python
# Find the highest Science marks
np.max(data[:,2])
```

Get details of the student who scored more than 90 in Math

```python
# Get details of the student who scored more than 90 in Math
data[data[:,1] > 90]
```
