### **Valid Sudoku**
- A Sudoku is valid if it fulfills following conditions:
    - It should have total matrix size of (9,9)
    - Every row and column should have unique elements between 1 to 9, such that no element repeats again
    - Every row should have sum of 45

```python
s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
```

```python
# Verify that if this is a valid Sudoku based on the condition given
```

```python
s.shape
```

```python
for row in s:
    if len(set(row)) != 9:
        print('Invalid Sudoku')
        break
    else:
        for col in s.T:
            if len(set(col)) != 9:
                print('Invalid Sudoku')
                break
else:        
    print('Valid rows and columns')
```

```python
for row in s:
    if len(set(row)) != 9:
        print('Invalid Sudoku')
        break

# print('Valid rows and columns')

for col in s.T:
    if len(set(col)) != 9:
        print('Invalid Sudoku')
        break
       
print('Valid rows and columns')
```

```python
r = np.sum(s, axis=1)
r == 45
```

```python
for i in r:
    if i != 45:
        print('Invalid Sudoku')
        break
else:
    print('Valid Sudoku')
```

```python
import numpy as np

# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

```python
#Get the shape of the matrix.
data.shape
```

```python
#Find the average age of students.
np.mean(data[:,0])
```

```python
#Extract Math marks of all students.
data[:,1]
```

```python
#Find the highest Science mark.
np.max(data[:,2])
```

```python
#Get details of the student who scored more than 90 in Math.
data[data[:,1] > 90]
```

```python
#Increase Math marks of all students by 5.
data[:,1] + 5
data
```
