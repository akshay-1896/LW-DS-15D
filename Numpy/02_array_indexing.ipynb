Numpy Indexing and Slicing of Vectors
(Same as of list in Python)

```python
arr = np.arange(11,21)
arr
```

```python
arr[3]
```

```python
arr[1:5]
```

```python
arr[:4]
```

```python
arr[1:]
```

```python
arr[3::]
```

```python
arr_slice = arr[2:9:2]
arr_slice
```

Numpy Indexing and Slicing of Matrix

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
arr[0] # Prints 0th row
```

```python
arr[5]
```

```python
arr[0,2]
```

```python
arr
```

```python
arr[5,4]
```

```python
arr[5,0:4]
```

```python
slice = arr[0:2,1:3]
slice
```

```python
arr[0:,1]
```

```python
arr[3:,3:]
```

```python
arr[:,4:]
```

```python
arr[:,1:2]
```

Boolean Indexing

```python
arr = np.arange(11,21)
arr
```

```python
# to get only even numbers

bool_index = arr % 2 == 0
bool_index
```

```python
arr = arr[bool_index]
arr
```
