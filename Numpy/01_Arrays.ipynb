### Vectors and Matrices
- Vectors are one-dimensional, formed through list
- Matrices are two-dimensional, formed through numpy array
- Multi-dimensional matrices are termed as Tensors

```python
l = [[1,2,3],[4,5,6],[7,8,9]] # vector
l
```

```python
np.array(l) # Now this becomes a matrix
```

Array Generation Function

```python
# Generating array from 1 to 10 with 2 steps
arr = np.arange(1,11,2)
arr
```

```python
arr = np.zeros(6) # Vector form
arr
```

```python
arr = np.zeros((4,8)) # 4 Rows and 8 columns
arr
```

```python
arr = np.ones((3,2))
arr
```

```python
#Linear Spacing -- Provides array in linearly spaced format within a certified range

arr = np.linspace(1,5,2)
arr
```

```python
arr = np.linspace(1,5,3)
arr
```

```python
arr = np.linspace(0,1,100)
arr
```
