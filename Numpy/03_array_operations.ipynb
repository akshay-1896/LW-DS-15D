Deep and Shallow copy

```python
a = np.arange(1,21)
a
```

```python
slice = a[:4]
slice
```

```python
slice = slice * 10 # Slicing takes a new different location in memory --> Deep Copy
slice
```

```python
a
```

```python
a[4:6] * 10
```

```python
a
```

```python
# b = a # both have same location in the memory --> Shallow Copy
b = a.copy() # --> Deep Copy
b
```

```python
b[0] = 99
b
```

### Advanced Array Manipulation

Stacking Array
- We cannot stack two matrices, because they are already stacked arrays

```python
A
```

```python
B
```

```python
# For stacking, we use vectors
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])
```

```python
#Vertical Stack
np.vstack((a,b)) # gives 2-D array
```

```python
# Horizontal Stack
np.hstack((a,b)) # gives 1-D array
```

```python
# Column stack
np.column_stack((a,b))
```

Splitting array
- Splits into 2-D matrix

```python
c = np.arange(16).reshape(4,4)
c
```

```python
# Horizontal Split
np.hsplit(c,2) # You need to have equal divisions for split
```

```python
np.hsplit(c,3) # Throws error because 4 cannot be divided into three parts
```

```python
np.vsplit(c,2)
```
